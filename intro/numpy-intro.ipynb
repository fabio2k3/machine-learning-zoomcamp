# Numpy


```python
import numpy as np
```

# Crear Array

```python
# Array de Ceros
np.zeros(10)
```

```python
# Array de unos
np.ones(10)
```

```python
# Array de un Numero
np.full(10, 2.5)
```

```python
# Crear array a paritr de lista
lis = [ 1,2,3,5,7,12]
a = np.array(lis)
print(a)
print(a[2]) # Indices igual que lista o array clasico
a[2] = 10
print(a)
```

```python
# Crear un array en un rango [ a, b)
np.arange(3,10)
```

```python
# Crear array con elementos con igual separacion entre ellos es decir misma diferencia entre el primero y el 2do y asi sucesivamente

arr1 = np.linspace(0,100,11)
arr2 = np.linspace(1,5,20)

print(arr1)
print(arr2)
```

```python
# Funciona igual con Matrices

matrix1 = np.zeros((5,2))
print(matrix1)

mat2 = np.array([[1,2], [2,3]])
print(mat2)
```

# Generar array aleatoriamente

```python
np.random.rand(5,2)
```

```python
np.random.seed(2)
np.random.rand(5,2) * 100
```

```python
# Random con enteros
np.random.seed(2)
np.random.randint(low=0, high = 100, size=(4,4))
```

# Operaciones con elementos del array (Númericas)


```python
a = np.arange(5)
print(a)
a += 1 
print(a)

b = np.arange(5)
print(b)
b *= 2
print(b)

# Así se puede hacer con varias operaciones
```

# Operaciones entre Array

```python
a = np.ones(5)
b = np.arange(5)

print(a + b)
```

# Operaciones binarias

```python
a = np.arange(10)
b = 2 * np.arange(10)

print(a)
c = (a >= 2)
print(c) 
```

```python
a = 1 + np.arange(10)
b = 2 +  np.arange(10)


print(a)
print(b)
d = (a >= b)
print(d)

a = a[(a>b) | (a%2 == 0)]
print(a)
```

# Otras operaciones

```python
a = np.arange(6)

print(a.min())
print(a.sum())
print(a.mean())
print(a.std())
```
