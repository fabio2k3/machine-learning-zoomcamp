# Cargar los Datos

```python
import pandas as pd

df = pd.read_csv("data.csv")
```

```python
# Observar las primeras 5 filas

df.head()
```

```python
# Depurar nombre de columnas

"""
Modificamos los nombres de las columnas para depurar y trabajar más cómodo:

1- Convertir todo minúscula
2- Remplazar espacios en blanco por '_'

"""
df.columns = df.columns.str.lower().str.replace(' ', '_')
```

```python
# Normalizar los valores de la columnas que sean Cadenas

strings = list(df.dtypes[df.dtypes == 'string'].index)
strings

for col in strings:
    df[col] = df[col].str.lower().str.replace(' ', '_')

df.head()
```

### Análisis Exploratorio de Datos (EDA)

```python
for col in df.columns:
    print(col)         # nombre de la columna
    print(df[col].unique()[:5])  # Algunos valores
    print(df[col].nunique())   # valores que son únicos
    print()
```
